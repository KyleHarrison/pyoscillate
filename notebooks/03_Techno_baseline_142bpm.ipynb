# Techno Baseline @ 142 BPM

A classic rolling techno bassline layered with an FM atmosphere pad, hats, and a slow-winding sub drone — all at 142 BPM.

Each voice ("patch") lives in its own module under `src/pysynth/patches/`. This notebook just builds a `Tempo`, asks each patch module for a `Patch` (a `sequencer` + `voice` pair), and starts/stops it. Edit a patch module and rerun its cell to hear the change — `autoreload` is enabled in the setup cell below, so no kernel restart needed. If a patch is currently playing, call its `.stop()` first so you don't layer the old version under the new one.

## Setup

```python
%load_ext autoreload
%autoreload 2

from pyo.lib.server import Server

from pysynth.tempo import Tempo
from pysynth.patches import atmosphere, bass, drone, hat, low_hat

s = Server().boot()
s.start()

tempo = Tempo(bpm=142)
```

## Bass

Rolling 16-step bassline through a resonant, LFO-swept lowpass filter. See `src/pysynth/patches/bass.py`.

```python
bass_patch = bass.build(tempo)
bass_patch.start()
```

## Atmosphere (FM pad + arpeggiator)

FM pad voice arpeggiated at 8th notes, with a slow amplitude swell and reverb. See `src/pysynth/patches/atmosphere.py`.

```python
atmosphere_patch = atmosphere.build(tempo)
atmosphere_patch.start()
```

## Hi-hat

Subtle high-passed noise tick, once per 8th note, for top-end texture. See `src/pysynth/patches/hat.py`.

```python
hat_patch = hat.build(tempo)
hat_patch.start()
```

## Low hat

Darker noise tick, once per quarter note, as a rarer, dubbier accent. See `src/pysynth/patches/low_hat.py`.

```python
low_hat_patch = low_hat.build(tempo)
low_hat_patch.start()
```

## Drone

Slow-winding FM drone: note changes once every 8 bars, with a continuously drifting timbre. See `src/pysynth/patches/drone.py`.

```python
drone_patch = drone.build(tempo)
drone_patch.start()
```

## Stop

Each patch can be stopped/restarted on its own by rerunning `<name>_patch.stop()` / `.start()`. This cell stops everything at once.

```python
bass_patch.stop()
atmosphere_patch.stop()
hat_patch.stop()
low_hat_patch.stop()
drone_patch.stop()
```
