# Techno Baseline @ 142 BPM

A classic rolling techno bassline layered with an FM atmosphere pad, hats, and a slow-winding sub drone — all at 142 BPM.

Each voice ("patch") lives in its own module under `src/pysynth/patches/`. This notebook builds a `Tempo`, a shared `Clock`, and a `PatchRack`, then asks each patch module for a `Patch` (a `sequencer` + `voice` pair) and starts it through the rack. Patches divide down from the one `Clock` rather than each running their own independent timer, so they land on the same downbeat even after you edit and rerun an individual patch cell — `autoreload` is enabled in the setup cell below, so no kernel restart needed.

Each patch cell goes through `rack.start("name", ...)` rather than calling `.start()` directly, so rerunning a cell after an edit always stops the previous version first automatically — you can't accidentally layer the old voice under the new one, even if you forget. If sound ever gets stuck that the rack doesn't know about (e.g. from code that bypassed it), see the "Panic reset" cell at the end, which restarts the audio server outright.

## Setup

```python
%load_ext autoreload
%autoreload 2

from pyo.lib.server import Server

from pysynth.clock import Clock
from pysynth.tempo import Tempo
from pysynth.patches import PatchRack, atmosphere, bass, clock_tick, drone, hat, low_hat

s = Server().boot()
s.start()

tempo = Tempo(bpm=142)
rack = PatchRack()

# single shared pulse every patch (except clock_tick, which is deliberately
# free-running) divides down from, so they lock to a common downbeat instead
# of just a common tempo - see src/pysynth/clock.py
clock = Clock(tempo)
clock.start()
```

## Bass

Rolling 16-step bassline through a resonant, LFO-swept lowpass filter. See `src/pysynth/patches/bass.py`.

```python
bass_patch = rack.start(
    "bass",
    bass.build(
        tempo,
        clock,
        root_freq=bass.ROOT_FREQ,
        filter_res=0.75,
        filter_base=900,
        filter_range=400,
    ),
)
```

## Atmosphere (FM pad + arpeggiator)

FM pad voice arpeggiated at 8th notes, with a slow amplitude swell and reverb. See `src/pysynth/patches/atmosphere.py`.

```python
atmosphere_patch = rack.start(
    "atmosphere",
    atmosphere.build(
        tempo,
        clock,
        arp_root=atmosphere.ARP_ROOT,
        fm_ratio=0.5012,
        fm_index=4,
        reverb_size=0.85,
        reverb_damp=0.6,
        reverb_bal=0.5,
    ),
)
```

## Hi-hat

Subtle high-passed noise tick, once per 8th note, for top-end texture. See `src/pysynth/patches/hat.py`.

```python
hat_patch = rack.start(
    "hat",
    hat.build(
        tempo,
        clock,
        cutoff_freq=hat.CUTOFF_FREQ,
        level=0.5,
        decay=0.2,
    ),
)
```

## Low hat

Darker noise tick, once per quarter note, as a rarer, dubbier accent. See `src/pysynth/patches/low_hat.py`.

```python
low_hat_patch = rack.start(
    "low_hat",
    low_hat.build(
        tempo,
        clock,
        cutoff_freq=low_hat.CUTOFF_FREQ,
        level=0.3,
        decay=0.5,
    ),
)
```

## Clock tick

Pink Floyd "Time"-style clock shop: four noise ticks, each on its own free-running period in real seconds rather than the tempo grid, so they drift past each other and never share a downbeat — a quiet background texture, not a groove element. See `src/pysynth/patches/clock_tick.py`.

```python
clock_tick_patch = rack.start(
    "clock_tick",
    clock_tick.build(
        tempo,
        level=clock_tick.OVERALL_LEVEL,
        wood_q=3,
        glass_q=6,
    ),
)
```

## Drone

Slow-winding FM drone: note changes once every 8 bars, with a continuously drifting timbre. See `src/pysynth/patches/drone.py`.

```python
drone_patch = rack.start(
    "drone",
    drone.build(
        tempo,
        clock,
        root_freq=drone.DRONE_ROOT,
        reverb_size=0.9,
        reverb_damp=0.3,
        reverb_bal=0.9,
    ),
)
```
