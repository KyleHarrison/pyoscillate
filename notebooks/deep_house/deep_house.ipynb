## Setup

Run this once at the start of the notebook to initialize the shared audio engine,
tempo, rack, and clock.

```python
%load_ext autoreload
%autoreload 2

from pyoscillate.clock import Clock
from pyoscillate.patches import PatchRack, setup_notebook
from pyoscillate.patches.drums.clap import clap as clap
from pyoscillate.patches.drums.hat import groove as hat
from pyoscillate.patches.drums.kick import kick as kick
from pyoscillate.patches.drums.percussion import percussion as percussion
from pyoscillate.patches.musical.chord import chord as chord
from pyoscillate.patches.tonal.bass import groove as bass
from pyoscillate.tempo import Tempo

# Start the shared audio engine once for this notebook.
# Replace `1` with the output device index reported by `pa_list_devices()`.
server = setup_notebook(output_device=1)

tempo = Tempo(bpm=122)
clock = Clock(tempo)
clock.start()
rack = PatchRack()
```

## Patch build debuggers

Each section below calls the real `build()` function, so autoreload picks up
saved changes in the patch module. Build, start, and stop are separate cells to
show which lifecycle stage fails. The live Flet GUI for this rack is
`src/flet/deep_house/app.py`.

## Kick - Round

Rounded, physical kick body with a soft transient click. Four-on-the-floor foundation. See `src/pyoscillate/patches/drums/kick/kick.py`.

```python
DEBUG_KICK_ROUND = "debug_kick_round"
rack.stop(DEBUG_KICK_ROUND)

debug_kick_round = kick.build(
    tempo,
    clock,
    style="round",
    level=0.62,
    drive=0.12,
    punch=1.0,
    length=1.0,
    click=1.0,
)
print(type(debug_kick_round.voice).__name__, sorted(debug_kick_round.controls))
```

```python
rack.start(DEBUG_KICK_ROUND, debug_kick_round)
```

```python
debug_kick_round.set("level", 0.7)
debug_kick_round.set("drive", 0.2)
# Run this when finished, before rebuilding the patch.
rack.stop(DEBUG_KICK_ROUND)
```
