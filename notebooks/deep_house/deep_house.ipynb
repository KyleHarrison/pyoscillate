## Setup

Run this once at the start of the notebook to initialize the shared audio engine,
tempo, rack, and clock.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

from IPython.display import display

from pyoscillate.clock import Clock
from pyoscillate.patches import PatchRack, setup_notebook
from pyoscillate.patches.tonal.bass import groove as bass
from pyoscillate.patches.musical.chord import chord as chord
from pyoscillate.patches.drums.hat import groove as hat
from pyoscillate.patches.drums.kick import kick as kick
from pyoscillate.patches.drums.percussion import percussion as percussion
from pyoscillate.patches.presets import PresetController, save_preset
from pyoscillate.tempo import Tempo

# Each JSON file in this folder stores one complete preset for all patch widgets.
CATALOG_DIR = Path("catalog")
preset_controller = PresetController(CATALOG_DIR)
display(preset_controller.display())

# Start the shared audio engine once for this notebook.
# Replace `1` with the output device index reported by `pa_list_devices()`.
server = setup_notebook(output_device=1)

tempo = Tempo(bpm=122)
clock = Clock(tempo)
clock.start()
rack = PatchRack()
```

```python
# Kick build debugger
#
# These cells call the real kick.build() function, so autoreload picks up
# saved changes in kick.py. Build, start, and stop are separate to show
# which lifecycle stage fails.
```

```python
DEBUG_KICK_NAME = "debug_kick"
rack.stop(DEBUG_KICK_NAME)

debug_patch = kick.build(
    tempo,
    clock,
    style="round",
    level=0.62,
    drive=0.12,
)
print(type(debug_patch.voice).__name__, sorted(debug_patch.controls))
```

```python
rack.start(DEBUG_KICK_NAME, debug_patch)
```

```python
debug_patch.set("level", 0.5)
debug_patch.set("drive", 0.2)
# Run this when finished, before rebuilding the patch.
rack.stop(DEBUG_KICK_NAME)
```
