## Setup

Run this once at the start of the notebook to initialize the shared audio engine,
tempo, rack, and clock.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

from IPython.display import display

from pyoscillate.clock import Clock
from pyoscillate.patches import (
    PatchRack,
    atmosphere,
    bass,
    clock_tick,
    drone,
    hat,
    low_hat,
    setup_notebook,
)
from pyoscillate.patches.presets import PresetController, save_preset
from pyoscillate.tempo import Tempo

# Each JSON file in this folder stores one complete preset for all patch widgets.
CATALOG_DIR = Path("catalog")
preset_controller = PresetController(CATALOG_DIR)
display(preset_controller.display())

# Start the shared audio engine once for this notebook.
# Replace `1` with the output device index reported by `pa_list_devices()`.
server = setup_notebook(output_device=1)

tempo = Tempo(bpm=132)
clock = Clock(tempo)
clock.start()
rack = PatchRack()
```

## Bass

Rolling 16-step bassline through a resonant, LFO-swept lowpass filter.
See `src/pyoscillate/patches/bass.py`.

```python
bass.widget(rack, tempo, clock, preset_controller)
```

## Atmosphere (FM pad + arpeggiator)

FM pad voice arpeggiated at 8th notes, with a slow amplitude swell and reverb.
See `src/pyoscillate/patches/atmosphere.py`.

```python
atmosphere.widget(rack, tempo, clock, preset_controller)
```

## Hi-hat

Subtle high-passed noise tick, once per 8th note, for top-end texture.
See `src/pyoscillate/patches/hat.py`.

```python
hat.widget(rack, tempo, clock, preset_controller)
```

## Low hat

Darker noise tick, once per quarter note, as a rarer, dubbier accent.
See `src/pyoscillate/patches/low_hat.py`.

```python
low_hat.widget(rack, tempo, clock, preset_controller)
```

## Clock tick

Pink Floyd "Time"-style clock shop: four noise ticks, each on its own free-running
period in real seconds rather than the tempo grid, so they drift past each other and never
share a downbeat. See `src/pyoscillate/patches/clock_tick.py`.

```python
clock_tick.widget(rack, tempo, preset_controller)
```

## Drone

Slow-winding FM drone: note changes once every 8 bars, with a continuously drifting timbre.
See `src/pyoscillate/patches/drone.py`.

```python
drone.widget(rack, tempo, clock, preset_controller)
```

## Stop

Each checkbox toggles its patch on or off. The dropdown at the top loads a complete
preset across all patches, and the save cell below writes the current widget values to the catalog.

```python
# Save the current widget values as a new preset when needed.
save_preset(CATALOG_DIR, "demo", preset_controller.values())
```
