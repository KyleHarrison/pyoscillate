## Setup

```python
%load_ext autoreload

%autoreload 2


from pathlib import Path


from IPython.display import display

from pyo.lib.server import Server


from pyoscillate.clock import Clock

from pyoscillate.tempo import Tempo

from pyoscillate.patches import PatchRack, atmosphere, bass, clock_tick, drone, hat, low_hat

from pyoscillate.patches.presets import PresetController, save_preset


from pyo.lib.server import pa_list_devices


# prints numbered input/output devices with channel counts; find your

# headphones in the output list and note its index

pa_list_devices()


# replace with the output device index printed above for your headphones

s = Server(nchnls=2)

s.setOutputDevice(1)

s.boot()

s.start()


tempo = Tempo(bpm=132)

rack = PatchRack()


# The catalog lives beside this notebook, so each JSON file is one complete

# set of values for all patch widgets.

CATALOG_DIR = Path("catalog")

preset_controller = PresetController(CATALOG_DIR)

display(preset_controller.display())


# single shared pulse every patch (except clock_tick, which is deliberately

# free-running) divides down from, so they lock to a common downbeat instead

# of just a common tempo - see src/pyoscillate/clock.py

clock = Clock(tempo)

clock.start()
```

## Bass



Rolling 16-step bassline through a resonant, LFO-swept lowpass filter. See `src/pyoscillate/patches/bass.py`.

```python
bass.widget(rack, tempo, clock, preset_controller)
```

## Atmosphere (FM pad + arpeggiator)



FM pad voice arpeggiated at 8th notes, with a slow amplitude swell and reverb. See `src/pyoscillate/patches/atmosphere.py`.

```python
atmosphere.widget(rack, tempo, clock, preset_controller)
```

## Hi-hat



Subtle high-passed noise tick, once per 8th note, for top-end texture. See `src/pyoscillate/patches/hat.py`.

```python
hat.widget(rack, tempo, clock, preset_controller)
```

## Low hat



Darker noise tick, once per quarter note, as a rarer, dubbier accent. See `src/pyoscillate/patches/low_hat.py`.

```python
low_hat.widget(rack, tempo, clock, preset_controller)
```

## Clock tick



Pink Floyd "Time"-style clock shop: four noise ticks, each on its own free-running period in real seconds rather than the tempo grid, so they drift past each other and never share a downbeat — a quiet background texture, not a groove element. See `src/pyoscillate/patches/clock_tick.py`.

```python
clock_tick.widget(rack, tempo, preset_controller)
```

## Drone



Slow-winding FM drone: note changes once every 8 bars, with a continuously drifting timbre. See `src/pyoscillate/patches/drone.py`.

```python
drone.widget(rack, tempo, clock, preset_controller)
```
