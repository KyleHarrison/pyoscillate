## Setup

Run this once at the start of the notebook to initialize the shared audio engine,
tempo, rack, and clock.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path

from IPython.display import display

from pyoscillate.clock import Clock
from pyoscillate.patches import PatchRack, setup_notebook
from pyoscillate.patches.presets import PresetController, save_preset
from pyoscillate.patches.psyambient import (
    bass_chaos,
    bass_drone,
    bass_rumble,
    mid_arp,
    mid_canon,
    mid_generative,
    soundscape_filter,
    soundscape_fm,
    soundscape_wash,
)
from pyoscillate.tempo import Tempo

# Each JSON file in this folder stores one complete preset for all patch widgets.
CATALOG_DIR = Path("catalog")
preset_controller = PresetController(CATALOG_DIR)
display(preset_controller.display())


# Start the shared audio engine once for this notebook.rack = PatchRack()

# Replace `1` with the output device index reported by `pa_list_devices()`.clock.start()

server = setup_notebook(output_device=1)clock = Clock(tempo)

tempo = Tempo(bpm=70)

# psyambient patches are mostly free-running rather than clock-locked, but# tempo/clock are kept around for any later patches that do want the grid.
```

## Soundscape - chaotic FM pad

Free-running FM pad whose timbre is driven entirely by `Rossler`/`Lorenz` chaotic
attractors rather than a clocked note pattern - dreamy, slowly wandering, never
quite repeating. See `src/pyoscillate/patches/psyambient/soundscape_fm.py`.

```python
soundscape_fm.widget(rack, preset_controller)
```

## Soundscape - filter-swept pad

A static harmonic-rich drone carved by a chaotically-swept resonant lowpass
filter - a more angular, "breathing" psychedelic filter-sweep character.
See `src/pyoscillate/patches/psyambient/soundscape_filter.py`.

```python
soundscape_filter.widget(rack, preset_controller)
```

## Soundscape - washy detuned pad

`SuperSaw` voice smeared with chorus, reverb, and delay for a shoegaze-style
dream-pop ambience - width and haze rather than timbral wander.
See `src/pyoscillate/patches/psyambient/soundscape_wash.py`.

```python
soundscape_wash.widget(rack, preset_controller)
```

## Mid - slow pentatonic arpeggio

Fixed pentatonic melody, gliding between notes every `step_bars` bars, phase-locked
to the shared clock. See `src/pyoscillate/patches/psyambient/mid_arp.py`.

```python
mid_arp.widget(rack, tempo, clock, preset_controller)
```

## Mid - generative melody

Free-running generative melody: a new pentatonic note drawn at random every few
seconds, in real time rather than locked to the shared clock. See
`src/pyoscillate/patches/psyambient/mid_generative.py`.

```python
mid_generative.widget(rack, preset_controller)
```

## Mid - two-voice canon

Two generative melodic voices on unrelated free-running periods, drifting in and
out of alignment like an ever-shifting call and response. See
`src/pyoscillate/patches/psyambient/mid_canon.py`.

```python
mid_canon.widget(rack, preset_controller)
```

## Bass - slow-swelling sub drone

Near-static low fundamental that breathes in and out in level rather than
changing pitch or timbre. See `src/pyoscillate/patches/psyambient/bass_drone.py`.

```python
bass_drone.widget(rack, preset_controller)
```

## Bass - chaotic sub drift

Near-static low fundamental whose pitch wanders unpredictably within a narrow
range, for an organic, unstable rumble. See
`src/pyoscillate/patches/psyambient/bass_chaos.py`.

```python
bass_chaos.widget(rack, preset_controller)
```

## Bass - textural noise rumble

Filtered `BrownNoise` blended with a faint sub sine, for an unpitched
"earthquake" low end rather than a tonal bass. See
`src/pyoscillate/patches/psyambient/bass_rumble.py`.

```python
bass_rumble.widget(rack, preset_controller)
```
